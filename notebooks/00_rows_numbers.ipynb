```python
files = ["train_triplets.csv", "test_triplets.csv", "val_triplets.csv"]

total_lines = 0
for file in files:
    path = f"../data/processed/{file}"  
    try:
        with open(path, "r", encoding="utf-8") as f:
            lines = sum(1 for _ in f)
            total_lines += lines
            print(f"{file}  contains {lines} rows")
    except FileNotFoundError:
        print(f"file {file}   is not found {path}")

print(f"total rows in all files {total_lines}")
```
